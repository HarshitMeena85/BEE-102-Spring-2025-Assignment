- **Input** : A FASTA file with multiline sequences. A header line (Starting with **'>'**) comes before the start of each sequence in the input file
- **Output** : A FASTA file with the multiline sequences converted to single line sequences. The format of the header line is conserved in the output file

**Steps :**
- Initialize an empty string `sequence` that stores the already read fasta line
- For each line in the input file, if the first element of the line equal `'>'` then it means that it's a header so we do the following
   - First if the string `sequence` is not empty then stores it as a new single line sequence in the output following the header line
   - Again make the `sequence` string to start storing next line
- Otherwise just add the line to the string `sequence` which strips the new line and convert the mutliline sequence into a single line sequence
- At the end if `sequence` is not empty, just add it to the output file with the header line

```python

with open('multiline_input.fasta', 'r') as infile, open('output.fasta', 'w') as outfile:
        sequence = ''
        for line in infile:
            line = line.strip()
            if line[0] == '>':
                if len(sequence) > 0:
                    outfile.write(sequence + '\n')
                    sequence = ''
                outfile.write(line + '\n')
            else:
                sequence += line
        if len(sequence) > 0:
            outfile.write(sequence + '\n')

print("The Multi-line FASTA file has been formatted into a Single line FASTA file and saved as 'output.fasta'.")
```
