- **INPUT** : A nucleotide sequence
- **OUTPUT** : A Transition matrix in the form of a dictionary

**Steps :**
- Initialize a python Dictionary to store the count of transition from A to B for A,B in {A,C,G,T} and (A ≠ B)
- Then to calculate the probability of transitioning from A to B divide the number of transitions from A to B by the total number of transitions from A 

```python
import pandas as pd
# Input DNA sequence
sequence = "CTTCATGTGAAAGCAGACGTAAGTCA"

# Initialize transition counts by making a dictionary of dictionaries
nucleotides = ['A', 'C', 'G', 'T']
transition_counts = {n1: {n2: 0 for n2 in nucleotides} for n1 in nucleotides}

# Count transitions
for i in range(len(sequence) - 1):
    current = sequence[i]
    next_nuc = sequence[i + 1]
    if current in nucleotides and next_nuc in nucleotides:
        transition_counts[current][next_nuc] += 1
    else :
        print(f"Invalid nucleotide found: {current} or {next_nuc}. Skipping this transition.")

# Normalize to get probabilities
transition_matrix = {n1: {} for n1 in nucleotides}
for n1 in nucleotides:
    total = sum(transition_counts[n1].values())
    for n2 in nucleotides:
        if total > 0:
            transition_matrix[n1][n2] = transition_counts[n1][n2] / total
        else:
            transition_matrix[n1][n2] = 0.0

# Print the transition matrix
df = pd.DataFrame(transition_matrix).T
print("Transition Matrix:")
display(df)
print("Rows = Current Nucleotide")
print("Columns = Next Nucleotide")
```
